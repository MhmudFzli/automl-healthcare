```python
import warnings
import pandas as pd

warnings.filterwarnings("ignore")
```

## Merge original dataset with discharge notes features

```python
discharge_htn_features_df = pd.read_csv("../data/processed/discharge_htn_features.csv")
final_hosp_dataset_df = pd.read_csv("../data/processed/final_hosp_dataset.csv")

# Use LEFT JOIN to keep all rows from the hosp dataset
final_hosp_discharge_df = pd.merge(
    final_hosp_dataset_df,
    discharge_htn_features_df,
    on='subject_id',
    how='left'   # 👈 keeps all subjects from final_hosp_dataset_df
)

print(final_hosp_dataset_df.shape)
print(final_hosp_discharge_df.shape)
final_hosp_discharge_df.head()
```

```python
final_hosp_discharge_df = final_hosp_discharge_df.drop(columns=['htn_first_mention_time', 'htn_last_mention_time'])
```

```python
print(final_hosp_discharge_df.shape)
final_hosp_discharge_df.head()
```

```python
missing_values_prop = final_hosp_discharge_df.isnull().mean()
a = missing_values_prop*100.0
print("missings proportion:\n", a)
```

```python
final_hosp_discharge_df.to_csv("../data/processed/final_hosp_discharge.csv", index=False)
```

## Merge original dataset with radiology notes features

```python
final_hosp_discharge_df = pd.read_csv("../data/processed/final_hosp_discharge.csv")
rad_compact_df = pd.read_csv("../data/processed/radiology_compact_summary.csv")

# Use LEFT JOIN to keep all rows from the hosp dataset
final_hosp_discharge_radio_df = pd.merge(
    final_hosp_discharge_df,
    rad_compact_df,
    on='subject_id',
    how='left'   # 👈 keeps all subjects from final_hosp_dataset_df
)

print(final_hosp_discharge_df.shape)
print(final_hosp_discharge_radio_df.shape)
final_hosp_discharge_radio_df.head()
```

```python
final_hosp_discharge_radio_df = final_hosp_discharge_radio_df.drop(columns=['rad_first_mention_time', 'rad_last_mention_time'])
```

```python
final_hosp_discharge_radio_df.to_csv("../data/processed/final_hosp_discharge_radio.csv", index=False)
```
